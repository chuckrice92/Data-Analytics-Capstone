# Exploratory Data Analysis

```python
sns.scatterplot(x='Average Meal Price', y='Revenue', data=df, hue='Location')

plt.xlabel("Meal Price")
plt.ylabel("Revenue")

plt.show()
```

```python
sns.histplot(x='Revenue',data=df, bins=20)

plt.xlabel("Revenue")
plt.ylabel("Count")
plt.show()
```

```python
sns.barplot(x='Cuisine', y='Revenue', data=df, hue='Location')

plt.xlabel("Cuisine")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.scatterplot(x='Seating Capacity', y='Revenue', hue='Cuisine', data=df)

plt.xlabel("Seating Capacity")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.scatterplot(data=df, x='Social Media Followers', y='Revenue')

sns.regplot(
    data=df,
    x="Social Media Followers",
    y="Revenue",
    scatter=False,       # only the line
    color="yellow",
    line_kws={"linewidth": 2.5},
    ci=95
)

plt.xlabel("Social Media Followers")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.scatterplot(data=df, x='Marketing Budget', y='Revenue')

sns.regplot(
    data=df,
    x="Marketing Budget",
    y="Revenue",
    scatter=False,       # only the line
    color="yellow",
    line_kws={"linewidth": 2.5},
    ci=95
)

plt.xlabel("Marketing Budget")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.barplot(x='Location', y='Revenue', data=df)

plt.xlabel("Location")
plt.ylabel("Revenue")
plt.show()
```

```python
sorted_order = df.sort_values('Revenue', ascending=False)['Cuisine']
sns.barplot(x='Cuisine', y='Revenue', data=df, order=sorted_order)

plt.xlabel("Cuisine")
plt.ylabel("Revenue")
plt.show()
```

```python


# Select only numeric columns for correlation calculation
numeric_df = df.select_dtypes(include=['number'])

# Create correlation heatmap with numeric data only
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', linewidths=0.5, linecolor='black', fmt='.2f')
plt.show()
```
