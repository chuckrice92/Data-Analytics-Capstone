```python
sns.scatterplot(x='Average Meal Price', y='Revenue', data=df, hue='Location')
plt.show()
```

```python
sorted_order = df.sort_values('Revenue', ascending=False)['Cuisine']
sns.barplot(x='Cuisine', y='Revenue', data=df, order=sorted_order)
```

```python


# Select only numeric columns for correlation calculation
numeric_df = df.select_dtypes(include=['number'])

# Create correlation heatmap with numeric data only
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', linewidths=0.5, linecolor='black', fmt='.2f')
plt.show()
```
